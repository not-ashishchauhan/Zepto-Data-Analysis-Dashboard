```python
# Remove duplicates
ratings.drop_duplicates(inplace=True)

# Check data types
ratings.info()

# Verify rating range
ratings['Prod_Rating'].describe()
ratings['Delivery/Service_Rating'].describe()
```

```python
df = orders.merge(customers,on='C_ID')
```

```python
df = df.merge(products,on='P_ID')
```

```python
df = df.merge(transactions,on='Or_ID')
```

```python
df = df.merge(ratings,on='Or_ID')
```

```python
df['Revenue'] = df['Qty'] * df['Price']
```

```python
df['Discount_Amount'] = (
    df['Revenue'] * df['Coupon_Discount']/100
)
```

```python
df['Final_Revenue'] = (
    df['Revenue'] - df['Discount_Amount']
)
```

```python
df['Final_Revenue'].sum()
```

```python
top_products = (
    df.groupby('PName')['Final_Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
```
